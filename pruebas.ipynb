```python
# Constante especial: '-e .' se escribe en requirements.txt para activar setup.py
HYPEN_E_DOT = '-e .'

def obtener_requisitos(ruta: str) -> list[str]: 
    """
    =======================================================================
    PROPÓSITO DE LA FUNCIÓN:
    Lee un archivo de texto (como 'requirements.txt') y devuelve una lista 
    limpia con los nombres de las librerías a instalar (ej: ['pandas', 'numpy']).
    
    PARÁMETROS:
    - ruta (str): La ruta o nombre del archivo de texto a leer.
    
    RETORNA:
    - list[str]: Una lista de textos con los nombres de las librerías limpias.
    =======================================================================
    """
    
    # 1. Creamos una lista vacía para guardar los requisitos
    requisitos = []
    
    # 2. 'with open' abre el archivo de forma segura.
    #    Al terminar este bloque, Python CIERRA el archivo automáticamente.
    with open(ruta) as file_obj:
        
        # 3. readlines() lee todas las líneas del archivo de golpe y las mete en una lista.
        #    ¡OJO!: Cada elemento incluirá el salto de línea '\n' al final.
        #    Ejemplo: ['pandas\n', 'numpy\n', 'seaborn\n', '-e .\n']
        requisitos1 = file_obj.readlines()
        print("1. Líneas leídas directamente (crudas con \\n):")
        print(requisitos1)
        
        # 4. List Comprehension: Recorremos cada elemento 'i' de la lista
        #    y le quitamos el '\n' usando .replace("\n", "")
        #    Ejemplo: ['pandas', 'numpy', 'seaborn', '-e .']
        requisitos = [i.replace("\n", "") for i in requisitos1]
        print("\n2. Líneas limpias (sin el \\n):")
        print(requisitos)
        
        # 5. Si '-e .' está en la lista, lo eliminamos.
        #    ¿Por qué? Porque '-e .' solo sirve para activar setup.py,
        #    no es una librería que pip pueda descargar de internet.
        if HYPEN_E_DOT in requisitos:
            requisitos.remove(HYPEN_E_DOT)
            print("\n3. Lista final sin '-e .' (lista para setup):")
            print(requisitos)
            
    # 6. Devolvemos la lista final limpia
    return requisitos


# =======================================================================
# PRUEBA DE LA FUNCIÓN (Puedes ejecutar esto para verla en acción):
# =======================================================================
# resultado = obtener_requisitos('requirements.txt')
# print("\nResultado devuelto por la función:", resultado)
```

```python
from dataclasses import dataclass
@dataclass
class DataIngestionConfig:
    train_data_path: str = os.path.join('artifacts', "train.csv")
    test_data_path: str = os.path.join('artifacts', "test.csv")
    raw_data_path: str = os.path.join('artifacts', "data.csv")


class DataIngestion:

    def __init__(self):
        self.ingestion_config = DataIngestionConfig()

    def initiate_data_ingestion(self):
        logging.info("1. Iniciando el proceso de Ingesta de Datos...")

        try:
            df = pd.read_csv('notebook/data/stud.csv')
            logging.info("2. Dataset leído exitosamente como DataFrame de Pandas.")

            os.makedirs(
                os.path.dirname(self.ingestion_config.train_data_path),
                exist_ok=True
            )

            df.to_csv(
                self.ingestion_config.raw_data_path,
                index=False,
                header=True
            )
            logging.info("3. Archivo con datos originales (raw data) guardado.")

            logging.info("4. Iniciando división de datos en Train y Test...")
            train_set, test_set = train_test_split(
                df,
                test_size=0.2,
                random_state=42
            )

            train_set.to_csv(
                self.ingestion_config.train_data_path,
                index=False,
                header=True
            )

            test_set.to_csv(
                self.ingestion_config.test_data_path,
                index=False,
                header=True
            )

            logging.info("5. Ingesta de datos completada con éxito total.")

            return (
                self.ingestion_config.train_data_path,
                self.ingestion_config.test_data_path
            )

        except Exception as e:
            raise CustomException(e, sys)


if __name__ == "__main__":
    obj = DataIngestion()
    obj.initiate_data_ingestion()
```

# **Tutorial 5-End To End ML Project-Data Transformation Implementation Using Pipelines**

```python

import sys
import os
from dataclasses import dataclass
import numpy as np 
import pandas as pd
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from src.exception import CustomException
from src.logger import logging
from src.utils import save_object


@dataclass
class DataTransformationConfig:
    """Ruta donde se guardará físicamente el objeto preprocesador serializado (.pkl)"""
    preprocessor_obj_file_path: str = os.path.join('artifacts', "preprocessor.pkl")


class DataTransformation:
    def __init__(self):
        # Conectamos la configuración de rutas
        self.data_transformation_config = DataTransformationConfig()

    def obtener_objeto_transformador_datos(self):
        """
        Crea la tubería (Pipeline) de transformación:
        - Tubería Numérica: Rellena con la mediana y estandariza los números.
        - Tubería Categórica: Rellena con la moda, convierte a One-Hot y escala.
        """
        try:
            logging.info("[data_transformation.py] 1. Definiendo columnas numéricas y categóricas...")
           
            # Columnas con números continuos (notas)
            numerical_columns = ["writing_score", "reading_score"]
            
            # Columnas con texto (categorías)
            categorical_columns = [
                "gender",
                "race_ethnicity",
                "parental_level_of_education",
                "lunch",
                "test_preparation_course"
            ]

            logging.info("[data_transformation.py] 2. Construyendo el Pipeline Numérico...")


            num_pipeline = Pipeline(
                steps=[
                    # Rellena valores vacíos con la mediana (el valor del centro)
                    ("imputer", SimpleImputer(strategy="median")),

                    # Estandariza: media = 0 y desviación estándar = 1
                    ("scaler", StandardScaler())
                ]
            )

            logging.info("[data_transformation.py] 3. Construyendo el Pipeline Categórico...")

            cat_pipeline = Pipeline(
                steps=[
                    # Rellena valores vacíos con el dato más frecuente (la moda)
                    ("imputer", SimpleImputer(strategy="most_frequent")),
                    # Convierte texto ("male", "female") en columnas de ceros y unos
                    ("one_hot_encoder", OneHotEncoder()),
                    # Escala las matrices dispersas sin centrar la media
                    ("scaler", StandardScaler(with_mean=False))
                ]
            )
            logging.info("[data_transformation.py] 4. Combinando ambas tuberías con ColumnTransformer...")
            
            preprocessor = ColumnTransformer(
                transformers=[
                    ("num_pipeline", num_pipeline, numerical_columns),
                    ("cat_pipeline", cat_pipeline, categorical_columns)
                ]
            )
            logging.info("[data_transformation.py] 5. Objeto preprocesador creado exitosamente.")
           
           
           #Retornamos el preprocesador 
            return preprocessor


        except Exception as e:
            raise CustomException(e, sys)


    # Alias en inglés para mantener 100% de compatibilidad con Krish Naik
    #colocamos un alias para que pueda ser llamada de dos maneras
    get_data_transformer_object = obtener_objeto_transformador_datos


    def iniciar_transformacion_datos(self, train_path: str, test_path: str):

        """
        Aplica el transformador a los archivos train.csv y test.csv,
        separa la variable a predecir (math_score) y guarda el preprocesador (.pkl).
        """


        try:
            logging.info("[data_transformation.py] 6. Leyendo archivos train.csv y test.csv...")

            #lectura de los archivos 
            train_df = pd.read_csv(train_path)
            test_df = pd.read_csv(test_path)
            logging.info("[data_transformation.py] 7. Obteniendo el objeto transformador...")


            preprocessing_obj = self.obtener_objeto_transformador_datos() 
            # La variable objetivo que queremos predecir (Target)


            target_column_name = "math_score"
            # -------------------------------------------------------------
            # Separamos en X (características de entrada) e Y (nota de mate)
            # -------------------------------------------------------------
            logging.info("[data_transformation.py] 8. Separando variables de entrada y objetivo...")
            
            # Conjunto de Entrenamiento (Train)
            input_feature_train_df = train_df.drop(columns=[target_column_name], axis=1)
            target_feature_train_df = train_df[target_column_name]
            # Conjunto de Prueba (Test)
            input_feature_test_df = test_df.drop(columns=[target_column_name], axis=1)
            target_feature_test_df = test_df[target_column_name]
            # -------------------------------------------------------------
            # Aplicamos la transformación (fit_transform en Train, transform en Test)
            # -------------------------------------------------------------
            logging.info("[data_transformation.py] 9. Aplicando transformaciones matemáticas a los datos...")
            
            # fit_transform: Aprende la media/escala de train y transforma
            input_feature_train_arr = preprocessing_obj.fit_transform(input_feature_train_df)
            # transform: Solo aplica lo aprendido en train sobre los datos de test (evita Data Leakage)
            input_feature_test_arr = preprocessing_obj.transform(input_feature_test_df)
            # Unimos las características transformadas con la columna objetivo usando np.c_
            train_arr = np.c_[
                input_feature_train_arr, np.array(target_feature_train_df)
            ]
            test_arr = np.c_[
                input_feature_test_arr, np.array(target_feature_test_df)
            ]
            logging.info("[data_transformation.py] 10. Guardando preprocessor.pkl en la carpeta artifacts...")
            save_object(
                file_path=self.data_transformation_config.preprocessor_obj_file_path,
                obj=preprocessing_obj
            )
            logging.info("[data_transformation.py] 11. Transformación de datos completada con éxito.")
            # Devolvemos las matrices procesadas y la ruta del preprocesador
            return (
                train_arr,
                test_arr,
                self.data_transformation_config.preprocessor_obj_file_path,
            )
        except Exception as e:
            raise CustomException(e, sys)



    # Alias en inglés para mantener compatibilidad con el resto del pipeline:
    initiate_data_transformation = iniciar_transformacion_datos
```
