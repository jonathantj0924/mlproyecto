```python
# Constante especial: '-e .' se escribe en requirements.txt para activar setup.py
HYPEN_E_DOT = '-e .'

def obtener_requisitos(ruta: str) -> list[str]: 
    """
    =======================================================================
    PROPÓSITO DE LA FUNCIÓN:
    Lee un archivo de texto (como 'requirements.txt') y devuelve una lista 
    limpia con los nombres de las librerías a instalar (ej: ['pandas', 'numpy']).
    
    PARÁMETROS:
    - ruta (str): La ruta o nombre del archivo de texto a leer.
    
    RETORNA:
    - list[str]: Una lista de textos con los nombres de las librerías limpias.
    =======================================================================
    """
    
    # 1. Creamos una lista vacía para guardar los requisitos
    requisitos = []
    
    # 2. 'with open' abre el archivo de forma segura.
    #    Al terminar este bloque, Python CIERRA el archivo automáticamente.
    with open(ruta) as file_obj:
        
        # 3. readlines() lee todas las líneas del archivo de golpe y las mete en una lista.
        #    ¡OJO!: Cada elemento incluirá el salto de línea '\n' al final.
        #    Ejemplo: ['pandas\n', 'numpy\n', 'seaborn\n', '-e .\n']
        requisitos1 = file_obj.readlines()
        print("1. Líneas leídas directamente (crudas con \\n):")
        print(requisitos1)
        
        # 4. List Comprehension: Recorremos cada elemento 'i' de la lista
        #    y le quitamos el '\n' usando .replace("\n", "")
        #    Ejemplo: ['pandas', 'numpy', 'seaborn', '-e .']
        requisitos = [i.replace("\n", "") for i in requisitos1]
        print("\n2. Líneas limpias (sin el \\n):")
        print(requisitos)
        
        # 5. Si '-e .' está en la lista, lo eliminamos.
        #    ¿Por qué? Porque '-e .' solo sirve para activar setup.py,
        #    no es una librería que pip pueda descargar de internet.
        if HYPEN_E_DOT in requisitos:
            requisitos.remove(HYPEN_E_DOT)
            print("\n3. Lista final sin '-e .' (lista para setup):")
            print(requisitos)
            
    # 6. Devolvemos la lista final limpia
    return requisitos


# =======================================================================
# PRUEBA DE LA FUNCIÓN (Puedes ejecutar esto para verla en acción):
# =======================================================================
# resultado = obtener_requisitos('requirements.txt')
# print("\nResultado devuelto por la función:", resultado)
```
